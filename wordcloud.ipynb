````python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter



nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
````

````python
text = """
```text
Fashion, music are creative forms of art that capture style, beauty, emotions, and unique moments in time.i love listening to music. music inspires me. I really want to travel around the world..travelling is good. i feel so carefree while travelling. Fashion includes outfits, models, makeup, accessories, trends, and design. Photography uses cameras, lighting, colors, and composition to create powerful images and tell stories. Videos, reels, film, and social media have transformed fashion and digital media. Every photograph reflects creativity, perspective, style, memories, and personal expression. Technology, editing, marketing, storytelling, and study also play an important role in modern fashion and photography.
```
"""


# Print original text
print(text)
````

````python
words = word_tokenize(text.lower())

print("\nTokenized words:")
print(words)
````

````python
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word.isalpha() and word not in stop_words
]

print("\nFiltered words:")
print(filtered_words)
````

````python
word_freq = Counter(filtered_words)

print("\nWord frequencies:")
print(word_freq)


# Create WordCloud
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white'
).generate_from_frequencies(word_freq)
````

````python
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
````
